### 🔬 Observation 4.1: True Division (`/`) vs Floor Division (`//`) vs Modulo (`%`)
- `/` always returns a `float` (e.g., `23 / 4` → `5.75`).
- `//` discards decimals and keeps only the quotient (`23 // 4` → `5`).
- `%` calculates the **remainder** (`23 % 4` → `3`).
- `**` is exponentiation (`2 ** 3` → `8`).



````python
candies = 23
children = 4

candies_per_child = candies // children
candies_leftover = candies % children

print(f"Each child gets: {candies_per_child} candies")
print(f"Leftover in the jar: {candies_leftover} candies")

````

### 🔬 Observation 4.2: Equality (`==`) vs Assignment (`=`) & The String Trap
- `=` puts data into a container.
- `==` asks: *"Are both sides equal in value?"* and returns `True` or `False`.
- `!=` asks: *"Are both sides different?"*
- Trap: `"5" == 5` evaluates to **`False`** because string `"5"` is not integer `5`!



````python
print("5 == 5    ->", 5 == 5)
print("5 == 2    ->", 5 == 2)
print("5 != 2    ->", 5 != 2)
print("'5' == 5  ->", "5" == 5)  # Type mismatch!

````

### 🔬 Observation 5.1: The 2 Core Rules of Expressions
1. **Rule 1:** An expression is anything that Python evaluates down to a **single value**.
2. **Rule 2 (The Assignment Test):** Anything that can legally sit on the right-hand side of `=` is an expression!

```python
x = (7 - 3) * 0.5   # (7 - 3) * 0.5 is an expression!
x = 10              # 10 is an expression!
```

- **Expression:** Produces a value (e.g., `5 + 3`, `x == 1`, `round(3.5)`).
- **Statement:** Executes an action (e.g., `print(x)`, `x = 5`).



````python
# Four Types of Expressions in Action:

# 1. Arithmetic Expression:
result_num = (10 + 5) * 2

# 2. Logical / Comparison Expression:
is_allowed = (result_num > 20 and result_num < 50)

# 3. String Expression:
full_greeting = "Hello, " + "Coders!"

# 4. Conditional (Ternary) Expression:
grade = "Pass" if result_num >= 25 else "Fail"

print("1. Arithmetic:", result_num)
print("2. Logical:", is_allowed)
print("3. String:", full_greeting)
print("4. Conditional:", grade)

````

### 🔬 Observation 5.2: Operator Precedence (PEMDAS)
Python resolves operations in strict hierarchy:
1. **P**arentheses `()` *(Highest priority)*
2. **E**xponents `**`
3. **M**ultiplication / **D**ivision / Floor / Modulo `*`, `/`, `//`, `%`
4. **A**ddition / **S**ubtraction `+`, `-`
5. Comparisons `==`, `!=`, `<`, `>`, `<=`, `>=`
6. Logical `not`, then `and`, then `or` *(Lowest priority)*

> **Golden Developer Rule:** *"When in doubt, use parentheses to make your intent obvious!"*



````python
# Without parentheses:
val_1 = 5 + 2 * 3 ** 2
# 3 ** 2 = 9
# 2 * 9  = 18
# 5 + 18 = 23
print("Result without parentheses:", val_1)

# With parentheses:
val_2 = (5 + 2) * (3 ** 2)
# (7) * (9) = 63
print("Result with parentheses:   ", val_2)

````

### 🔬 Observation 6.1: Implicit vs. Explicit Conversion
- **Implicit:** Python auto-promotes an `int` to a `float` during arithmetic (`10 + 3.14` → `13.14`).
- **Explicit:** You convert explicitly using functions: `int()`, `float()`, `str()`, `bool()`.
- **Trap:** `int("hello")` raises a `ValueError` because `"hello"` contains letters, not digits!



````python
# Implicit auto-promotion:
auto_val = 10 + 3.14
print("10 + 3.14 produces:", auto_val, "which is type:", type(auto_val))

# Explicit casting:
score_str = "95"
score_num = int(score_str)
print("Converted score:", score_num + 5)

````
